#**isinstance(object, classinfo)**
Parameters
```
object → The variable or value you want to check.

classinfo → A class, type, or a tuple of classes/types to check against.
```
Return Value
```
True → If object is an instance of classinfo (or a subclass).
False → Otherwise.
```
#Examples

````python
# Example 1: Single type check
x = 5
print(isinstance(x, int))       # True

# Example 2: Multiple type check
y = [1, 2, 3]
print(isinstance(y, (list, tuple)))  # True (because y is a list)

# Example 3: Subclass check
class Animal:
    pass

class Dog(Animal):
    pass

d = Dog()
print(isinstance(d, Dog))       # True
print(isinstance(d, Animal))    # True (Dog is a subclass of Animal)
````

Difference between isinstance (1) and type (2):

Feature
1. isinstance(object, classinfo)
2. type(object) == classinfo

Checks subclasses
1. ✅ Yes — returns True if the object is an instance of the class or any subclass
2. ❌ No — returns True only if the object’s type matches exactly

Multiple types
1. ✅ Yes — can pass a tuple of types/classes
2. ❌ No — must check each type separately

Preferred for OOP
1. ✅ Yes — works well with inheritance
2. ❌ No — ignores inheritance

Readability

1.	✅ More descriptive for type checking
2.  ⚠️ Less clear for multiple checks

Example
1. isinstance(dog, Animal) → True if dog is Animal or subclass
2. type(dog) == Animal → False if dog is subclass

````python
class Animal:
    pass

class Dog(Animal):
    pass

d = Dog()

# isinstance() allows subclass match
print(isinstance(d, Animal))  # True

# type() requires exact match
print(type(d) == Animal)      # False
````

#**Polymorphism**
The literal meaning of polymorphism is the condition of occurrence in different forms. It refers to the use of a single type entity (method, operator or object) to represent different types in different scenarios.

In simple words, Polymorphism helps the code of python behave in different way in different case.

#Type of Polymorphism
1. Method Overloading: A class can have multiple methods with the same name but different parameters, and the method to be called is determined based on the number and types of arguments passed during the function call.
2. Method Overriding: Method overriding occurs when a subclass defines a method with the same name and parameters as a method in its superclass. When this occurs, the method in the subclass overrides the method in the superclass, allowing the subclass to provide its own implementation of
the method.
3. Operator Overloading: Python allows operators to be overloaded, so that they can be used with
user-defined classes. For example, the "+" operator can be overloaded to perform concatenation
on two string objects.
4. Duck Typing: In Python, an object's suitability for an operation is determined by its behavior (i.e.,
its methods and attributes) rather than its type. So, if two different objects have the same
behavior, they can be used interchangeably

#Built-in Polymorphism
* Function Polymorphism (len()):

    The len() function can accept
    strings, lists, tuples, or dictionaries,
    adapting its behavior to count
    elements based on the data type.

    ```
    print(len("Python"))
    print(len([1, 2, 3]))
    ```

* Operator Polymorphism (+):

    The + operator performs addition on numeric
    types but switches to concatenation when used
    with strings or lists.
    
    ```
    print(5 + 10)
    print("Hello " + "AI")
    ```
    
# **Runtime Polymorphism (Method Overriding)**
Runtime polymorphism occurs
when a subclass provides its own
specific implementation of a
method that is already defined in
its parent class. Python resolves
these method calls dynamically at
runtime based on the object
calling it

#Example:


````python
class PaymentGateway:
    def process_payment(self, amount):
        print("Loading....")



class CreditCard(PaymentGateway):
    def process_payment(self, amount):
        super().process_payment(amount)
        return f"Processing ${amount} via Credit Card authentication.\n"


class PayPal(PaymentGateway):
    def process_payment(self, amount):
        super().process_payment(amount)
        return f"Processing ${amount} via PayPal secure token."


# Unified interface function
def execute_checkout(payment_method, amount):
    print(payment_method.process_payment(amount))


# Usage
card = CreditCard()
wallet = PayPal()

execute_checkout(card, 250)
execute_checkout(wallet, 250)
````

````python
# syntactic sugar

a = 2
b = 4
print(a+b)


print("\n\n")
print(int.__add__(a,b))
````

#📝 What’s happening step by step:

You set a = 2 and b = 4.

When you write print(a + b), Python internally calls a special method named __add__.

So a + b is just a shortcut (syntactic sugar) for int.__add__(a, b). Both give the same result: 6.

#💡 Key idea:  
Syntactic sugar means a simpler, more human‑friendly way of writing something that could otherwise be written in a longer, more technical form.
```
a + b → easy to read.

int.__add__(a, b) → the actual function Python runs behind the scenes.
```

**👉 In short: the + operator is just a nicer way to call the __add__ method.**

# **Duck Typing**
Duck typing is a concept related to dynamic
typing where the type or class of an object is
less important than the methods it defines.
The name comes from the saying: “If it walks
like a duck and quacks like a duck, then we treat it as a duck.”

In Python, this means we don’t care about the actual type of the object.

As long as the object has the method we’re calling (fly() in this case), it works.

👉 Here, both Duck and Airplane have a fly() method.
So fly_test(entity) can accept either one — Python doesn’t check the class type, it just checks if the method exists.

# Example:

````python
class Duck:
  def swim(self):
    return "Duck swimming"

  def fly(self):
    return "Duck flying"

class Airplane:
  def fly(self):
    return "Airplane flying"
def fly_test(entity):
    print(entity.fly())


# Create objects
duck = Duck()
airplane = Airplane()

# Same function works with different objects
fly_test(duck)
fly_test(airplane)
````

This shows duck typing in action: one function works with different objects because they share the same method name.


# **Operator overloading**
Operator overloading is a feature in
Python that allows the same operator
to have different meanings according
to the context. For instance, the +
operator performs arithmetic addition
on numbers but concatenation on
strings.

````python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y
  def __add__(self, other):
    return Point(self.x + other.x, self.y + other.y)

  def __str__(self):  #this formats the code to be pretty else it would have shown something like <__main__.Point object at 6516456385235623>
    return f"Point({self.x}, {self.y})"


# Create Point objects
p1 = Point(1, 2)
p2 = Point(3, 4)

# Using the + operator on Point objects
p3 = p1 + p2
print(p3)
````

In this example, we’ve
overloaded the + operator for
the Point class by
implementing the __add__
special method.

**Deeper analysis**
When you write p1 + p2, Python automatically looks for a special method called __add__ inside the Point class.

The first object (p1) becomes self, and the second object (p2) becomes other.

here __add__ method says: “Take my x and y, add them to the other object’s x and y, and return a new Point.”

That’s why p1 + p2 creates a new Point(4, 6).

#For printing:

Normally, printing an object shows something ugly like <__main__.Point object at 0x...>.

By defining __str__, you tell Python how to display the object in a nice human‑friendly way.

So instead of the default, it prints Point(4, 6).

#🗣️ In simple words
+ → Python secretly calls __add__.

The second object automatically becomes the other argument.

__str__ → makes printing look nice.

Python does the object.__add__ call behind the scenes, and __str__ is just for friendly output.
