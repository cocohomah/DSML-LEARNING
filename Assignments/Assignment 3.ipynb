```python
from functools import reduce
students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
marks = [85, 42, 78, 91, 35, 67]
grade2 = []


#Display Serial Number and Name
def display(data = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]):
 data.sort()
 print("Serial No. Name")
 for i,v in enumerate(data):
    print (i,"        ",v)

#Combine and Display: Name & Marks
def combine(data1 = ["Ram  ", "Sita ", "Hari ", "Gita ", "Shyam", "Rita "],data2 = [85, 42, 78, 91, 35, 67]):
 new = list(zip(data1,data2))
 print("Name  Marks")
 for i,k in new:
    print(i," ",k)



#calculates the grade and remarks
def grade1(marks):
 if  marks > 90 and marks <= 100:
  return "Grade: A+ \n Remarks: Outstanding"
 elif marks > 80 and marks <= 90:
  return "Grade: A \n Remarks: Excellent"
 elif marks > 70 and marks <= 80:
  return "Grade: B+ \n Remarks: Very Good"
 elif marks > 60 and marks <= 70:
  return "Grade: B \n Remarks: Satisfactory"
 elif marks > 50 and marks <= 60:
  return "Grade: C+ \n Remarks: Acceptable"
 elif marks > 40 and marks <= 50:
  return "Grade: C \n Remarks: Basic"
 else:
  return "Grade: NG \n Remarks: Not Graded"


#Displays Name Marks Grade and Remarks
def grade(grade2,students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"],marks = [85, 42, 78, 91, 35, 67]):
 grade2 = map(grade1,marks)


 grade3 = zip(students,marks,grade2)
 for i,j,k in grade3:
    print("\n Name:",i,"\n Marks:",j,"\n" ,k)

#for passing students print
def passing(students = ["Ram ", "Sita", "Hari", "Gita", "Shyam", "Rita"],marks = [85, 42, 78, 91, 35, 67]):
  dat = list(zip(students,marks))
  newdat = list(filter(lambda x: x[1] > 40,dat))
  print("The passing student with their marks are: ")
  print("Name  Marks")
  for i,k in newdat:
    print(i," ",k)

#for adding marks
marks1 = lambda x,y: x+y

#for finding the sum of total marks
def sum_marks(marks = [85, 42, 78, 91, 35, 67]):
  result = reduce (marks1,marks)
  print("The total marks obtained by all students combined is: \n",result)


def maximum(a,b):
 if a[1]>b[1]:
  return a
 else:
  return b

#for finding the highest marks
def highest(students = ["Ram ", "Sita", "Hari", "Gita", "Shyam", "Rita"],marks = [85, 42, 78, 91, 35, 67]):
  dat = list(zip(students,marks))
  newdat = list(reduce(maximum,dat))
  print("The highest marks among all student is of : ", newdat[0],"\n& The highest marks is: ",newdat[1])


print("Welcome To My Program\n\n")

print("Enter 1 to Display all student names with their serial numbers starting from 1")
print("Enter 2 to Combine the student names and marks into a single collection containing student records.")
print("Enter 3 to Generate grades for each student")
print("Enter 4 to Create a list containing only the students who have passed the examination.")
print("Enter 5 to Calculate the total marks obtained by all students combined.")
print("Enter 6 to Find the highest mark among all students")
print("Enter 0 to exit the menu")


num = input("\n")
print("\n")
while num != "1" and num != "2" and num != "3" and num != "4" and num != "5" and num != "6" and num != "0":
    print("Invalid Input\n")
    num = input("Enter again: \n")
    print("\n")
if num == "1" :
  display()
elif num == "2":
  combine()
elif num == "3":
  grade(grade2)
elif num == "4":
  passing()
elif num == "5":
  sum_marks()
elif num == "6":
  highest()
else:
  print("Exiting The Menu...")



con = input("\nDo you want to continue? (yes/no): \n")
print("\n")
while con.lower().strip() != "yes" and con.lower().strip() != "no":
  print("Invalid Input \nPlease Enter again!\n")
  con = input("Do you want to continue? (yes/no): \n")
  print("\n")
if con.lower().strip()=="no":
        print("Thank you for using the program")
else:
  while con.lower().strip() == "yes":
   print("\n\nEnter 1 to Display all student names with their serial numbers starting from 1")
   print("Enter 2 to Combine the student names and marks into a single collection containing student records.")
   print("Enter 3 to Generate grades for each student")
   print("Enter 4 to Create a list containing only the students who have passed the examination.")
   print("Enter 5 to Calculate the total marks obtained by all students combined.")
   print("Enter 6 to Find the highest mark among all students")
   print("Enter 0 to exit the menu")


   num = input("\n")
   print("\n")
   while num != "1" and num != "2" and num != "3" and num != "4" and num != "5" and num != "6" and num != "0":
    print("Invalid Input\n")
    num = input("Please Enter again: \n")
    print("\n")
   if num == "1" :
    display()
   elif num == "2":
    combine()
   elif num == "3":
    grade(grade2)
   elif num == "4":
     passing()
   elif num == "5":
    sum_marks()
   elif num == "6":
    highest()
   else:
    print("Exiting The Menu...")


   con = input("\nDo you want to continue? (yes/no): \n")
   while con.lower().strip() != "yes" and con.lower().strip() != "no":
    print("Invalid Input \nPlease Enter again!\n")
    con = input("Do you want to continue? (yes/no): \n")
    print("\n")
   if con.lower().strip()=="no":
        print("\nThank you for using the program")
        break
```
