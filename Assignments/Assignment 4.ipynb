# **Mini Project: Student Management System**

## **Project Title:**

### ***Student Management System Using OOP Concepts***

**Problem Statement:**

Develop a Student Management System using Python that demonstrates
the concepts of Class, Object, Encapsulation, Inheritance, and Abstraction.
The system should allow users to store student information, calculate
grades, and display student details based on different student categories.


**Requirements**
Create an Abstract Class Student

       

> The abstract class should contain:




  ```



          1.   Student ID
          2.   Student Name
          3.   A private attribute__marks
          4.   A method to set marks
          5.   A method to get marks
          6.   An abstract method calculate_grade()
          7.   A method display_details()


  ```


````python
from abc import ABC, abstractmethod

class Learner(ABC):
    def __init__(self, student_id, full_name, score):
        self.student_id = student_id
        self.full_name = full_name
        self.__score = score
        self.result = ""

    def fetch_score(self):
        return self.__score

    def update_score(self, score):
        if 0 <= score <= 100:
            self.__score = score
        else:
            print("Marks must be between 0 and 100.")

    @abstractmethod
    def grade_calc(self):
        pass

    def show_info(self):
        print(f"\nStudent Id: {self.student_id}\nName: {self.full_name}\nMarks: {self.__score}\nGrade: {self.result}\n")


class DataStudent(Learner):
    def __init__(self, student_id, full_name, score):
        super().__init__(student_id, full_name, score)

    def grade_calc(self):
        marks = self.fetch_score()
        if marks >= 85:
            self.result = "A"
        elif marks >= 65:
            self.result = "B"
        elif marks >= 45:
            self.result = "C"
        elif marks >= 32:
            self.result = "D"
        else:
            self.result = "Fail"
        return self.result


class WebStudent(Learner):
    def __init__(self, student_id, full_name, score):
        super().__init__(student_id, full_name, score)

    def grade_calc(self):
        marks = self.fetch_score()
        if  marks > 90 and marks <= 100:
            self.result = "A+"
        elif marks > 80 and marks <= 90:
            self.result = "A"
        elif marks > 70 and marks <= 80:
            self.result = "B+"
        elif marks > 60 and marks <= 70:
            self.result = "B"
        elif marks > 50 and marks <= 60:
            self.result = "C+"
        elif marks > 40 and marks <= 50:
            self.result = "C"
        else:
            self.result = "NG"
        return self.result


print("Welcome To My Program")
while True:
    print("\n--- Student Management System ---")
    print("1. Add Data Science Student")
    print("2. Add Web Development Student")
    print("3. Exit")

    choice = input("Enter your choice (1-3): \n")

    if choice == "1":
        nm = input("Enter the name of student in Data Science: \n")
        sid = input(f"Enter the student id of {nm}: \n")
        sc = int(input(f"Enter the marks obtained by {nm}: \n"))
        stud = DataStudent(sid, nm, sc)
        stud.grade_calc()
        stud.show_info()

    elif choice == "2":
        nm = input("Enter the name of student in Web Development: \n")
        sid = input(f"Enter the student id of {nm}: \n")
        sc = int(input(f"Enter the marks obtained by {nm}: \n"))
        stud = WebStudent(sid, nm, sc)
        stud.grade_calc()
        stud.show_info()

    elif choice == "3":
        print("Exiting Student Management System...")
        print("Thank You For Using My Program")
        break

    else:
        print("Invalid choice. Please enter 1, 2, or 3.")
````
