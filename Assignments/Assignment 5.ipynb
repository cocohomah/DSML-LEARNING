# **Polymorphism Practice Questions**
* Create a base class Vehicle with a method move(). Create Car, Bike, and Plane
classes that implement their own version of move(). Write a function
start_journey(vehicle) that works with any vehicle object.
* Create a base class PaymentGateway with a method process_payment(amount).
Create CreditCard, PayPal, and Esewa classes that implement their own payment
processing logic. Write a function checkout(payment_method, amount) that works
with any payment object.
* Create a base class Shape with a method area(). Create Rectangle, Circle, and
Triangle classes that calculate their areas differently.
* Create a base class Employee with a method calculate_salary(). Create
FullTimeEmployee and PartTimeEmployee classes that calculate salary using
different formulas.
* Create classes PDFDocument, WordDocument, and ExcelDocument. Each class
should have an open_file() method. Write a function open_document(doc) that
opens any document type.
* Create a base class Account with a method calculate_interest(amount). Create
SavingsAccount, CurrentAccount, and FixedDepositAccount classes with different
interest calculations

```python
class Vehicle:
  def move(self):
    print("Vehicle is moving")

class Car(Vehicle):
  def move(self):
    print("Car is moving")

class Bike(Vehicle):
  def move(self):
    print("Bike is moving")

class Plane(Vehicle):
  def move(self):
    print("Plane is moving")

def start_journey(vehicle):
  vehicle.move()

car = Car()
bike = Bike()
plane = Plane()

start_journey(car)
start_journey(bike)
start_journey(plane)
```

```python
class PaymentGateway:
  def process_payment(self, amount):
    print(amount,"Payment is being processed")

class CreditCard(PaymentGateway):
  def process_payment(self, amount):
    print(amount,"Credit card payment is being processed")

class PayPal(PaymentGateway):
  def process_payment(self, amount):
    print(amount,"PayPal payment is being processed")

class Esewa(PaymentGateway):
  def process_payment(self, amount):
    print(amount,"Esewa payment is being processed")

def checkout(payment_method, amount):
  payment_method.process_payment(amount)
  print("Payment is successful")

credit_card = CreditCard()
paypal = PayPal()
esewa = Esewa()

checkout(credit_card, 1000)
print("\n\n")
checkout(paypal, 2000)
print("\n\n")
checkout(esewa, 3000)
```

```python
class shape:
  def area():
    pass
class rectangle(shape):
  def area(self, length, breadth):
    return length*breadth

class circle(shape):
  def area(self, radius):
    return 3.14*radius*radius

class triangle(shape):
  def area(self, base, height):
    return 0.5*base*height

def area(shape,*args):
  print( shape.area(*args))

rectangle = rectangle()
circle = circle()
triangle = triangle()

area(rectangle, 10, 20)
area(circle, 10)
area(triangle, 10, 20)
```

```python
class Employee:
  def calculate_salary(self):
    pass

class FullTimeEmployee(Employee):
  def calculate_salary(self, salary):
    return salary*12

class PartTimeEmployee(Employee):
  def calculate_salary(self, salary):
    return salary*6

def calculate_salary(employee, salary):
  return employee.calculate_salary(salary)

full_time_employee = FullTimeEmployee()
part_time_employee = PartTimeEmployee()

print(calculate_salary(full_time_employee, 1000))
print(calculate_salary(part_time_employee, 1000))
```

```python
class PDFDocument:
    def open_file(self):
        return "Opening PDF document"

class WordDocument:
    def open_file(self):
        return "Opening Word document"

class ExcelDocument:
    def open_file(self):
        return "Opening Excel document"

# Duck typing function
def open_document(doc):
    print(doc.open_file())

# Create objects
pdf = PDFDocument()
word = WordDocument()
excel = ExcelDocument()

# Same function works with different objects
open_document(pdf)
open_document(word)
open_document(excel)
```

```python
class Account:
  def calculate_interest(self, amount):
    pass

class SavingsAccount(Account):
  def calculate_interest(self, amount):
    return amount*0.05

class CurrentAccount(Account):
  def calculate_interest(self, amount):
    return amount*0

class FixedDepositAccount(Account):
  def calculate_interest(self, amount):
    return amount*0.02

def calculate_interest(account, amount):
  print( account.calculate_interest(amount))

savings_account = SavingsAccount()
current_account = CurrentAccount()
fixed_deposit_account = FixedDepositAccount()

calculate_interest(savings_account, 1000)
calculate_interest(current_account, 1000)
calculate_interest(fixed_deposit_account, 1000)

```
