```python

def student_result(student_data,name,marks,i):
 if marks >= 80:
           grade = 'A'
 elif marks >= 60 and marks <=79 :
           grade = 'B'
 elif marks >= 40 and marks <=59:
           grade = 'C'
 else:
           grade = 'F'
 student_data.update({'student'+ str(i) : {
          'name':name.strip(),
          'marks':marks,
          'grade' :grade
        }})




marks_list = []
student_data = { }
print("Enter the Data of 5 Students \n")
for i in range (5):
  name = str(input("Enter the name of student : \n"))
  marks = int(input("Enter the marks of student : \n"))
  student_result(student_data,name,marks,i)
  marks_list.append(marks)



print ("\n",student_data)
print("The highest Marks is", max(marks_list))
print("The Lowest Marks is", min(marks_list))
avg = sum(marks_list)/len(marks_list)
print("The Average Marks is", avg)
```

```python
def shopping_cart(product_data,name,price,total,i):
    product_data.update( {'product'+str(i):{
      'name': name.strip(),
      'price': price,}
    })



product_data = {}
total = 0
print("Enter the Data of 5 Products \n")
for i in range (5):
  name = input("Enter the name of product : \n")
  price = int(input("Enter the price of product : \n"))
  shopping_cart(product_data,name,price,total,i)
  total += price

if total >= 5000:
  discount = 0.15
elif total >= 3000 and total <= 4999:
  discount = 0.10
elif total >= 1000 and total <= 2999:
  discount = 0.05
else:
  discount = 0
discount_amt = total * discount
payable_amt = total - discount_amt

print ("\n",product_data)
print ("The Total Amount is: ", total)
print ("The Discount Amount: ", discount_amt)
print ("The Final Amount is: ", payable_amt)
```
