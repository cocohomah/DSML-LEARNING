````python
#upper() --> Makes string upper case
coupon_code = input ("Enter your coupon code \n")
if coupon_code.upper() == "DIS50" :
  print("Your coupon is applied")
else:
  print("Coupon is invaild. Please Try again!")

#lower() --> Makes string lower case
email = input ("Enter your email \n")
print ("Your email is: ", email.lower())
````

````python
#capitalize() --> Capitalize the first word of string
#title()      --> Capitalize the first word of every string
name = input ("Enter your name \n")
project_name = input ("Enter your project name \n")


print ("Your name is: ", name.capitalize(),"& your project is: ", project_name.title())
````

````python
#strip()  --> Removes whitespace from left and right
#lstrip() --> Removes whitespace from left
#rstrip() --> Removes whitespace from right
#while in bracket we can keep specific thing to remove from left and right
#if we want to

name = input ("Enter you name: \n")
hashtag = input ("Enter your favourtie trend: \n")
feeling = input ("Enter your feeling about it: \n")

print("Name: ", name.strip())
print("Favourite Trend: ", hashtag.lstrip("#"))
print("Trend's Feeling: ", feeling.rstrip("!"))
````

````python
#replace(old,new) --> Replaces old value with new one
Salary = input("Enter your annual salary: \n")
print ("Your annual salary is: ", Salary.replace("USD", "US Dollar"))
````

````python
#split(separator) --> makes string a list
#where the seperator in brackets defines what seperates the data lin list
#join(iterable)  --> joins list with separator and makes a string
#where syntax for join is actually separator.join(list_variable)
name = input ("Enter your list of fruits: \n")
lst = name.split(" ")
print("You have", len(lst), "fruits in your list")
print("They are: ", ", ".join(lst))
````

````python
#find(sub)--> finds the index of the string where sub is substring to be found
#while it gives -1 as value if the substring is not found
dat = input ("Enter your full file name:\n")
pos = dat.find(".pdf")
if pos != -1 :
  print("File type supported")
else:
  print("File type not supported")
````

# ***Syntax for exception handling (try and except)***
**try:**

    # Code that might cause an error

**except Some_Error_Type:**
   
    # Code that runs if that specific error happens


# **How it works**
**try: block** ``` → Python runs the code inside here. ```


*   If no error occurs → it skips the except and continues normally.

*    If an error occurs → Python immediately jumps out of the try block.


**except: block**` → This runs only if an error happens in the try.`

`In this block can specify the error type (like ValueError, TypeError, etc.).`



*   If the error matches, this block runs.

*   If it doesn’t match, Python keeps looking for another except if there is any, else crashes the program.

````python
#startswith(sub) --> checks if string starts with substring
#it returns boolean value 1 (true) or 0 (false)
a = input("Enter your professional name: \n")
a = a.upper()
if a.startswith("ER") == 1:
  print("You are a Engineer")
else:
  print("You are not a Engineer")
````

````python
#endswith(sub) --> checks if string ends with substring
#it returns boolean value 1 (true) or 0 (false)
dat = input ("Enter your full file name:\n")
if dat.endswith(".pdf") == 1 :
  print("File type is PDF")
else:
  print("File type is not PDF")
````

````python
#isalnum() --> checks if all characters are alphanumeric (number/alphabet/both)
#it returns boolean value 1 (true) or 0 (false)



#this program checks if entered symbol no is correct where
#it will be correct if its number or alphabet or both but
#will be incorrect if it has whitespace and symbols like @ # ! etc

dat = input ("Enter your symbol number: \n")
if dat.isalnum() == 1 :
  print("Valid symbol number")
else:
  print("Invalid symbol number")
````

````python
#isdigit() --> checks if all characters are numeric (numbers)
#it returns boolean value 1 (true) or 0 (false)

dat = input ("Enter your phone number: \n")
if dat.isdigit() == 1 :
  if len(dat) == 10:
    print("Correct phone number")
  else:
    print("Incorrect phone number")
else:
  print("Incorrect phone number")
````

````python
#isalpha() --> checks if all characters are alphabets
#it returns boolean value 1 (true) or 0 (false)

dat = input ("Enter your first name: \n")
if dat.isalpha() == 1 :
    print("First Name Accepted")
else:
  print("Incorrect Name")
````

````python
#islower() --> checks if all characters are lowercased
#isuppper() --> checks if all characters are uppercased
#they return boolean value 1 (true) or 0 (false)

dat1 = input ("Enter the link: \n")
if dat1.islower() == 1 :
    print("Link Accepted")
else:
  print("Incorrect Link: it should be lowercased")

dat2 = input ("Enter referral code: \n")

if dat2.isupper() == 1 :
    print("Referral Code Accepted")
else:
  print("Incorrect Referral Code: it should be uppercased")
````

````python
#isspace() --> checks if all characters are whitespaces
#they return boolean value 1 (true) or 0 (false)

dat1 = input ("Enter your review: \n")
if len(dat1) == 0 :
    print("Review cannot be empty")
elif dat1.isspace() == 1 :
   print("Cant leave space as review")
else:
  if dat1.isdigit() == 1 :
     print("Invalid Review")
  else:
    print("Review Accepted")
````
