#**if condition syntax**
```
if ___ condition ___:
the thing if its true
elif ___ condtion ___:
if its true
else:
if condition is not true
```



````python
age = int(input("enter your age \n"))
if age < 18 :
  print ("age is less than 18")
elif age == 18:
  print ("age is 18")
elif age > 18 and age < 40 :
  print ("age is greater than 18 and less than 40")
elif age == 40:
  print ("age is 40")
else:
  print ("age is greater than 40")
````

````python
age = int(input("enter your age \n"))
citizenship = input("enter country of your citizenship \n")
if age >= 18 and citizenship.upper().strip() == "NEPAL":
  print ("You can vote")
else:
  print ("You can not vote")
````
