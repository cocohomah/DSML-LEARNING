```python
#slicing in list
my_list = [0,1,2,3,4,5,6,7,8]
print(my_list[1])
print(my_list[1:5])
print(my_list[1:5:2])
print(my_list[:3])
print(my_list[::2])
print(my_list[:5:2])
print(my_list[3::2])
print(my_list[::-1]) #reverse the list
print(my_list[1:-2:1])
```

```python
lst = [1,2,3]
lst.append(6767) #takes and adds data last index
print("1st ",lst)
lst.extend([1,2,3,2]) #takes and adds data of 1 list to another
                    #list last index
print("2nd ",lst)
lst.insert(2,"hehe") #adds data in specific index syntax
                      #list_name.insert(index,data)
print("3rd ",lst)
lst.remove(3) #removes a first appearing value which is given in argument
print("4th ",lst)
lst.remove(3)
print("look the 3 which appeared first is removed so if i do it again another 3 after that one is removed \n",lst)
lst.pop(2) #removes the value of index given in argument
print("5th ",lst)
a = lst.index(6767) #gives the index of the value given in argument
print("6th ",a)
b = lst.count(2) #gives the number of repetion of the value in argument
print("7th ",b)
lst.sort() #sorts the list in ascending order
print("8th ",lst)
c = lst.copy() #copy the list so here lst is coppied in c (makes shallow copy)
print("9th ",c)
lst.reverse() #reverse the list
print("10th ",lst)
lst.clear() #clears the list
print("11th ",lst)
```

```python
from collections import deque #what deque does is make double ended queues
                              #where lists are used as queues
lst = deque(["elina","stefan","damon"])
print(lst)
lst.append("klaus")
print(lst)
lst.popleft()
print(lst)
```

```python
import copy
lst = [[1,2,3,4],[3,4,5]]
d = copy.deepcopy(lst)
c = lst.copy()
lst [0] [1] = 0


print("deep copy wala \n ",d)
print("original wala \n ",lst)
print("shallow copy wala \n ",c)

#this deepcopy and shallowcopy is only applicable in nested (nested list)
#in shallow copy the value is changed when changed in orginal
#but in deep copy the value is not changed when changed in orginal
```
