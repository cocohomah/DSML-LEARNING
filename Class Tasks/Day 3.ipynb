```python
#creation of tuple variable = (data1, data2,....)
dat = (1,2,3,4)
type (dat)
print (dat)
```

```python
#creation of tuple using constructor --> tuple() --> it takes only 1 argument
dat = tuple ((1,2,"hari",[1,3,4],5.6,(1,2,3)))
type (dat)
print (dat)
```

```python
#tuple function
dat = (1,2,3,4)
dat1 = (1,2,3,0,4,"",)
print(len(dat))
print(max(dat))
print(min(dat))
print(sum(dat))
print(any(dat1))#returns 1/true if any one value is non zero/non empty string
print(all(dat1)) #returns 1/true if all value is non zero or non empty string
print(sorted(dat))

```

```python
#indexing and getting values
dat = (1,3,4)
a = dat[0]
b = dat[1]
c = dat[2]
print (a,b,c)

#unpacking and getting values
x,_,z = dat #here values goes to the variable and "_" is used to skip the
            #data where data goes to "_" and is not used
print(x,z)

#slicing
print(dat[0::2]) #slicng variable[start:end:step]

#concatination
dat1 = (5,2,0)
dat2 = dat + dat1
print(dat2)

#repetition --> repeats tuple multiple times using *
dat3 = dat2*3
print (dat3)

#membership in tuples --> checks if values exist in tuple
print(2 in dat2)

#comparision --> compares element by element using comparison operators
print(dat<dat1) #here first element is compared in 0 index
print(dat[1]<dat1[1])


#we can delete whole tuple using del
del dat3
try:
  print(dat3)
  print("data is not deleted")
except NameError:
  print("data is deleted and it got NameError")

#but we cant delete individual element
try:
 del dat2[0]
 print("data is not deleted")
except TypeError:
  print("data is deleted and it got TypeError")
```

```python
#Creation of set variable = {value1, value2....}
#null set and dictionary is same so variable = {} is dictionary
datset={1,2,3,4}
datset1 = {}
print(type(datset))
print(type(datset1))

#creation of set using set constructor set()
datset2 = set ((1,2,5)) #takes only one argument & curly bracket can also be used
print(datset2)

datset3 = datset - datset2 #set difference like A-B
print(datset3)

datset3 = datset | datset2 #set union like A U B
print(datset3)


datset3 = datset & datset2 #set intersection like A n B
print(datset3)

datset3 = datset ^ datset2 #set symmetric difference like A ^ B
print(datset3)

datset3 = datset <= datset2 #Checks if its subset like A C B
print(datset3)

datset3 = 2 in datset2 #Checks if elements exist in set
print(datset3)
```
