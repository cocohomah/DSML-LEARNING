```python
def large(a,b,c):
 if a > b and a > c:
  return a
 elif b > a and b > c:
  return b
 else:
  return c


a = int (input ("Enter the first number \n"))
b = int (input ("Enter the second number \n"))
c = int (input ("Enter the third number \n"))
d = large(a,b,c)
print ("The largest number is \n",d)
```

```python
def vow(word):
  n = 0
  for i in range (0,len(word)):
    if word[i].lower() == 'a' or word[i].lower() == 'e' or word[i].lower() == 'i' or word[i].lower() == 'o' or word[i].lower() == 'u':
      n = n + 1
  return n
word = input("Enter the word \n")
num = vow(word)
print("the number of vowels is: ", num)
```

```python
def vow(word):
  n = 0
  for i in range (0,len(word)):
    if word[i].lower() in 'aeiou':
      n = n + 1
  return n
word = input("Enter the word \n")
num = vow(word)
print("the number of vowels is: ", num)
```

```python
def add (n):
  lst = []
  for i in range (0,n):
    a = input("Enter a number: \n")
    lst.append(a)
  file = open("numbers2.txt" , mode = "w")
  for i in range (len(lst)):
    file.write(lst[i]+"\n")
  file.close()

n = int(input("Enter the number of elements: \n"))
add(n)
```

```python
#filter() function
#it checks the elements with the given functions condition
def even(num):
 return num % 2 == 0
lst = [1,2,3,4,5]
even_lst = list(filter(even,lst))
print(even_lst)
```

```python
#reduce function
#it takes first 2 elements gives to function then
#takes result and next element and gives to function and keeps repeating this step
#till elements is ended
from functools import reduce


def mux(a,b):
  return (a*b)
lst1 = [1,2,3,4,5]
resdata = reduce(mux, lst1)
print (resdata)
```

```python
def maximum(a,b):
 if a>b:
  return a
 else:
  return b

lst = [1,5,2,8,99]
result = reduce(maximum, lst)
print (result)
```
