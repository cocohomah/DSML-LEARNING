```python
#Lamda Function
# syntax
# Function object = lambda argument: function_statement (single expression)
#where result is stored in function object
#no name so anonymus function
#used for short task
#example
sum = lambda a,b : a+b
print(sum(1,2))
```

```python
#square function using lambda
sqr = lambda a: a**2
print(sqr(3))
```

```python
#map and lambda
num = [1,2,3,4,5,6]
sqr = lambda a : a**2
print(list(map(sqr,num)))
```

```python
#filter and lambda
num = [1,2,3,4,5,6]
even = lambda a : a%2==0
print(list(filter(even,num)))
```

```python
#reduce and lambda
from functools import reduce

mun = lambda a,b: a+b

lst = [1,5,2,8,99]
result = reduce(mun, lst)
print (result)
```

```python
num = int(input("Enter a number: \n"))
odd_even = lambda num : "Even" if num%2==0 else "Odd"
print(odd_even(num))
```

```python
students = [
("Ram", 80),
("Hari", 95),
("Sita", 85)
]

#here sort sees the list elements one by one and since there is tuples
#our function makes sort see the "1" index in the tuple to sort,
#the key = MY_FUNCTION helps sort to work according to function
#and the reverse place is there in sort like
#list_name.sort(reverse = True/False, key = MY_FUNCTION)
students.sort(reverse= True, key=lambda x: x[1]) #list.sort(reverse=True|False,key=myFunc)

print(students)
```

```python
import pprint #this will help me print dictionary in mannered way
              #since they are in nested form
employees = [
{"name": "A", "salary": 50000},
{"name": "C", "salary": 30000},
{"name": "B", "salary": 70000} ]

#here sort sees the list elements one by one and since there is dictionary
#our function makes sort see the "salary" key in the dictionary to sort,
#the key = MY_FUNCTION helps sort to work according to function
#and the reverse place is there in sort like
#list_name.sort(reverse = True/False, key = MY_FUNCTION)

employees.sort(key=lambda emp: emp["salary"])
pprint.pprint(employees)

print("\n")


employees.sort( key=lambda emp: emp["name"])

pprint.pprint(employees)
```

```python
#This is list comprehension
data = ["welcome", "to", "world"]
is_even = lambda data : [word.upper() for word in data if 'e' in word]
print(is_even(data))
```

```python
data = [1,2,3,4,5,6]
print(["even"if num%2==0 else  "odd" for num in data ])
```

```python
#squaring if its not number then adding in list
numbers = [1, 2, 3, 4, 5, 6]

result = {num: num ** 2 for num in numbers if num % 2 == 0}

print(result)
```

```python
#checking if number is odd or even
data = [1,2,3,4,5,6]
print({num : "even"if num%2==0 else  "odd" for num in data })
```
