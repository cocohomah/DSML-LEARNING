# **RECURSION FUNCTION**

````python
#Factorial Number
def fact(n):
  if n == 0 or n ==1:
    return 1
  else:
    return n*fact(n-1)



print("The Factorial is:",fact(5))
````

````python
#Fibonacci Number
def fibo(n):
  if n <= 1:
    return n
  else:
    return (fibo(n-1)+fibo(n-2))

n = int(input("Enter number of terms of fibonacci:\n"))
print("The Fibonacci is:")
a = []
for i in range(n):
 a.append(fibo(i)) #could do print(fact(i)) too

print(a)
````

# **Scope and the LEGB Rule**
* The scope of a name defines the area of a program in which you can unambiguously access that
name, such as variable names, functions, or any other object.
* Python looks for a name in the following scopes, if they exist, in this order:  
  1. Local Scope
  2. Enclosing Scope
  3. Global Scope
  4. Built-in Scope

* The global and built-in scopes are always present. However, a local scope is only present when
referencing a name within a function definition and an enclosing scope is only present if you define
an inner function.

# **Decorators**
What is Decorators?

 * Decorators are simply a function which takes function as argument and enhances the base function without modyifing it.


Creating Decorators

**Syntax:**

```
def Decorator_name(Base_Function): =>here base function is named locally
                                   so name can be anything and function
                                   is passed here as argument

  def Enhanced_function(arguments):
    statements
    Base_Function(Argument)    =>calling of function
    statements
  return Enhanced_Function => here the function is returned after enhancement

```
Applying Decorators

**Syntax:**

Method 1:
```
new_base_function = Decorator_name(old_base_function)
new_base_function()
```

Method 2:
```
old_base_function = Decorator_name(old_base_function)
old_base_function()
```

Method 3:
```
 =>this (@) is going to wrap the function/ Enhance the function till the
 =>def (defination) of one function is ended

@Decorator_name   
def old_base_function():
 print("This is enhanced using decorator using @ ")

old_base_function()
```


**NOTE: IN ALL THESE EXAMPLE FROM ABOVE WE JUST WRITE NAME OF FUNCTION {WITHOUT USING ()}WHICH PASSES FUNCTION / FUNCTION DEFINATION RATHER THAN CALLING THE FUNCTION WHICH IS DIFFRENT THING**

````python
#EXAMPLE OF DECORATOR
import time
from datetime import datetime, timedelta

def timer_dec(base_fn):
  def enhanced_fn(*args, ** kwargs):
    start_time = time.time() #starts time
    result = base_fn(*args, ** kwargs)
    end_time = time.time() #stops time
    print(f"Task time: {end_time - start_time} seconds \n") #show the difference in time / execution time
    return result
  return enhanced_fn

@timer_dec
def brew_tea(tea_type, steep_time, milk):
  print(f"Brewing {tea_type} tea ... ")
  time.sleep(steep_time)
  print("Tea is ready!")

@timer_dec
def make_matcha():
  print("Making matcha ... ")
  time.sleep(1)  #delays the process
  print("Matcha is ready!")
  return f"Drink matcha by {datetime.now() + timedelta(minutes=30)}" #says when to eat by adding 30 minutes in current time


brew_tea("green", steep_time=1, milk = "oat")
print(make_matcha())
````

````python
#ok simple decorator with return value

def greet(func):
  def hello(name,age1):
    print(f"Hello {name}")
    func(name,age1)
  return hello

@greet
def age(name,year):
  print(f"Your age is {year}")


age("Dhiraj",21)
````

````python
#another way to do it using *args and **Kwargs but simple is

def logger(func):
  def wrap(*args,**kwargs):
    print(f"Calling {func.__name__}")
    func(*args,**kwargs)
    print(f"Finished {func.__name__}")
  return wrap


@logger
def greet(name,age):
  print(f"Hello {name}, Age: {age}")

greet("Dhiraj",21)
````

# **Generators**

**Generator objects are created by using yeild inside a function.**
<br> <br>
Lets say we need one value at a time but dont want to use all the value at the same time and are convserving the memory then return wont work since it gives single value and forgets all the operation/value for future use.
<br> <br>
But when we use generator it says the function to give that value for now and remember the state or value of function such that when we later call it, it should start from the current state/value to give us a new value
<br> <br>

`yield` **means "pause here and return a value, but remember the current state of the function."**
<br> <br>
**Syntax For Creating a Generator:**

```
def Function_name(arguments):
 statements..
 yeild value_to_return_and_remember
```
<br> <br>
**Syntax For Retriving Generator Values:**

```
Function_name(arguments)  => it gives the first value
Function_name(arguments)  => second value and so on
```
or
```
a = Function_name(arguments)
next(a)
```

````python
#EXAMPLE

def  greet(num): #creating generator's function

  yield (f"hello {num}") #using yield to return as generator

for i in range (10): #running loop from 0-9
  a = greet(i)       #rertiving value of generator and storing in a
  print(next(a))     #printing value
````


# **Basic Difference Between Iterator and Generator**
Iterator: We can create an iterator using the iter() function or by defining the iter() and next() methods in a
class.

Generator: We create a generator using a function that contains the yield keyword.
Generator uses the yield keyword and automatically saves its local variables and execution state between
function calls.

Generators help us write cleaner, shorter, and more readable code compared to custom iterator classes.
Both iterators and generators are memory efficient because they generate values one at a time instead of
storing all values in memory. However, generators are generally easier to implement.

Key Point: Every generator is an iterator, but not every iterator is a generator.
