#Method Overriding
Method overriding is a feature in object-oriented programming that allows a subclass to provide a specific implementation of a method that is already defined in its superclass. This is useful for runtime polymorphism, where the method to be executed is determined by the object that invokes it.

Basically, Method overriding is an object-oriented programming feature where a child class provides a specific way to run a method already defined in its parent class. It uses the same method name, parameters, and return type to change or customize how that action works for the child object

```python
#example of Method Overriding
#without super
class Animal:
    def make_sound(self):
        print("The animal makes a sound.")

class Dog(Animal):
    def make_sound(self):
        print("The dog barks.")

obj = Dog()
obj.make_sound()
```

```python
#with super
class Animal:
  def make_sound(self):
    print("The animal makes a sound.")

class Dog(Animal):
  def make_sound(self):
    super().make_sound()
    print("The dog barks.")

obj = Dog()
obj.make_sound()
```

```python
#another example of modifying the methods of parent class using the super

class Employee:
  def __init__(self, nm, sal):
    self.name = nm
    self.salary = sal
  def getName(self):
    return self.name
  def getSalary(self):
    return self.salary


class SalesOfficer(Employee):
  def __init__(self, nm, sal, inc):
    super().__init__(nm, sal)   #super() is used to call a method from the parent class.
    self.incentives = inc        #Here, super().__init__(nm, sal) calls the Employee
                                #constructor to initialize name and salary, avoiding the
                                #need to write the same code again in the child class.

  def getSalary(self):
    return self.salary + self.incentives

e1 = Employee("Rajesh", 9000)
print("Total salary for {} is Rs {}".format(e1.getName(), e1.getSalary()))
s1 = SalesOfficer("Kiran", 1000, 1000)
print("Total salary for {} is Rs {}".format(s1.getName(), s1.getSalary()))
```

# **Abstraction in Python**
Abstraction is the process of hiding implementation details and exposing only the essential functionality to the user. It is used to hide the
implementation details from the user and expose only necessary parts, making the code simpler and easier to interact with.

An abstract class is a class that cannot be instantiated, but can be subclassed. It defines a set of abstract methods that must be implemented by its
subclasses. These abstract methods define the interface of the class and are used to enforce a contract between the abstract class and its subclasses.

### **It focus on what an object does not how it does it**

```python
from abc import ABC, abstractmethod #importing class and decorator

class Animal(ABC): #ABC = AbstractBaseClass, and Animal is inheriting ABC properties

    @abstractmethod  #decorator for making an abstract method
    def make_sound(self):  #the method which is there but doesnt work when we directly call it by making its own object
                           #but can be inherited in other subclass/child class to use this method
        print('hello')


class Dog(Animal):
    def make_sound(self):
      super().make_sound() #abstraction used while inheriting above function using super
      print("Bark")


class Cat(Animal):
    def make_sound(self):  #abstraction used while not inheriting above function using super
        print("Meow")

try:
  obj = Animal()
  obj.make_sound()
except Exception as e:
  print("Error is Given : ",e)
  print("\nlook how it can't be used/called directly after making its instance we need subclass or child class to make it work\n\n")


dog = Dog()
dog.make_sound()

print("\n\n")

cat = Cat()
cat.make_sound()
```

# **Access Modifiers in Python: Encapsulation**
At times, we might need to restrict access to certain variables or methods while
programming. In such situations, we use access modifiers in Python encapsulation that limit the access of the class data members from outside the class. They help protect the internal data of a class and its instances.
Encapsulation in Python supports three types of access modifiers:

* Public Members: Accessible from anywhere, both inside and outside the class.
* Protected Members: Accessible within the class and its subclasses. They are defined
with a single underscore prefix (_).
* Private Members: Accessible only within the class. They are defined with a double underscore prefix (__).

```python
class Example:

    def __init__(self):
        self.public_var = 1          # Public variable
        self.__private_var = 2       # Private variable
        self._protected_var = 3      # Protected variable

    def public_method(self):
        print("Public Method")
        print("Private Variable:", self.__private_var)

    def __private_method(self):
        print("Private Method")

    def _protected_method(self):
        print("Protected Method")


# Creating object
obj = Example()

# Accessing public members
print(obj.public_var)
obj.public_method()

# Accessing protected members (possible, but not recommended)
print(obj._protected_var)
obj._protected_method()
```

```python
# Accessing private members directly (will cause error)
print(obj.__private_var)
obj.__private_method()
```
