```python
#sum of 10 numbers
a = 1
sum = 0
while a <= 10 :
  sum += a # it means sum = sum + a
  a += 1 # it means a = a + 1
print ("the sum of first 10 number is: ",sum)
```

```python
#break statement use
#sum of n numbers
a = 1
sum = 0
while True :
  sum += a # it means sum = sum + a
  a += 1 # it means a = a + 1
  b = input("do you want to continue? y/n ")
  if b == "n":
    break #breaks the loop if program sees break

print ("the sum of number is: ",sum)
```

```python
a = 0
while a < 10 :

  a += 1
  if a == 5 :
    continue # skips the loop code if the program sees continue
  print(a)
```

```python
a = 0
while a < 10 :

  a += 1
  if a == 5 :
    pass #Does nothing at all, then moves to the next line
         #usually used where code is needed but dont wanna keep anything
  print(a)
```

```python
#take 5 number and add numbers in list and find sum of positive numbers only
sum = 0
lst = []
for i in range (0,5):
  b = int (input ("enter a number: \n"))
  if b >= 0:
    lst = lst + [b] #concatenation
    sum = sum + b
print (" the list is: \n", lst, "\n and the sum is: ", sum)
```

```python
#take 5 number and add numbers in list and find sum
#of positive and negative number seperately
sum1 = 0
sum2 = 0
pos_lst = []
neg_lst = []
for i in range (5):
  b = int (input ("enter a number: \n"))
  if b >= 0:
    pos_lst = pos_lst + [b] #concatenation
    sum1 = sum1 + b
  else:
    neg_lst = neg_lst + [b] #concatenation
    sum2 = sum2 + b
print (" the positive list is: \n", pos_lst, "\n and the of positive numbers sum is: ", sum1)
print (" the negative list is: \n", neg_lst, "\n and the of negative numbers sum is: ", sum2)
```

```python
#atm system
amt = 10000


while True:
  print("Enter A for Checking Balance")
  print("Enter B for Depositing Cash")
  print("Enter C for Exiting \n")
  c = input()
  if c.upper() == "A":
    print("\nYour Balance is:\n",amt,"\n")
  elif c.upper() == "B":
    d = int(input("\nEnter the amount you want to deposit:\n"))

    amt += d
    print("\nYour New Balance is:\n",amt,"\n")
  else:
    break
```
