# **The __init __() Method**
All classes have a built-in method called `__init__()`, which is always executed when the class is being initiated.

The `__init__() `method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.


#**The self Parameter**
The `self` parameter is a reference to the current instance of the class.

It is used to access properties and methods that belong to the class.

````python
class Person:
  def __init__(self, name, age): #init method to assign properties like name and age to class
                                 #Self refers to itself in a class
    self.name = name
    self.age = age

p1 = Person("Emil", 36) #giving value Emil and 36 as properties value

print(p1.name)
print(p1.age)
````

#**FIRST: Lets learn about the ways we can add attribute into the object**
Here when we create and instance of an object we can add attribute in two ways

**Hardcoded way or global way:**
```
class object_name:
   pass
instance_name = object_name()
instance_name.attribute_name = value
```

**While creating instance or using __init__ or dynamically:**
```
class object_name:
   def __init__(self,name):
     self.name = name

instance_name = object_name(value)
```
lets talk about this more, at first case we dont have any attribute or function taking value so we just hardcode it in global scope for adding attribute to the instance.

In Second case, we used `init` function which asks for the value at the time of creation of instance which is passed in global scope as a argument. After that the value is there locally in the `init` which means it is still not creating a attribute in the instance (which is why error is showing in the first one)
so in-order to create a attribute in instance at the same time of creation of an instance we have done :
```
 self.name = name
 ```
 As discussed above `self` takes the instance as argument and just above code makes an attribute for name and stores the `name` which was passed as value/argument while creating a instance due to which the attribute is created for that instance. Which is equivalent to:
 ```
 instance_name.attribute_name = value
 ```

 So, after that when we run the program the attribute is created and the value is stored as name so we can access it later.

There might be question if two is doing for all why to even use one

The answer is lets say we have item class and 2 items (phone and laptop) so we might say this laptop has physical numpad but phone might not have physical numpad so in that case we can hard code it. The example of this case is below.


````python
#example
class items:
  def __init__(self,name):
    self.name = name

phone = items("Phone")
print(phone.name)
print("\n")
laptop = items("Laptop")
print(laptop.name)

laptop.Physical_numpad = True
print("\n")
if laptop.Physical_numpad:
  print("Laptop has physical numpad")
else:
  print("Laptop doesn't have physical numpad")
````

So this means if we use `init` contructor and already assigned attribute it doesnt mean we can assign more attributes

````python
#Example of Second case
class Fruits:
  def __init__(self,name):
    self.name = name

apple = Fruits("Apple")
print(apple.name)
````

#**What does __ init __(self) does or How does it works?**
Two condition are set:


1.   If there is `init` function defined by programmer while creating instance:
   
    In this case, python looks for init function and finds it so while creating a instance it execute or calls that function.
    Since the function is called it takes all the argument required and execute all the code which are present and needed to
    run as normal function and after the completion of the code the object is created at the same.

    So in simple words, it creates the instance and perfoms the function task at same time of creation of instance.

2.   If there is not init function:

    In this case, python looks for init function but it doesn't find the init function. Hence it uses its default/Built-in
    init function with not more argument than self and execute it due to which we dont see any error.

 **The function code it execute in 2nd case is given below:**
 ```
def __init__(self):
    pass
```


**Lets talk about self more**

lets say we need a method/function to do something when we call it in an instance and also we need to pass argument so we can do by two ways

First:
```
class Fruits:
   def greet(self,name):
      print(f"I am {name}")

apple = Fruits()
apple.greet("apple")
```
Here, when we run it we created a instance then while calling the method we passed name locally as argument of method and we need to pass it everytime we want to call it.

Second:
```
class Fruits:
   def __init__(self,name):
     self.name = name
   def greet(self):
      print(f"I am {self.name}")

apple = Fruits("apple")
apple.greet()
```
As i have said `self` is just a name and object's instance is passed onto it.

Here, we created the instance while passing the name during creation and while being created the passed name is used as value while creating the attribute `self.name`. Afterthat, when we call the method `greet` there is `self` as mandatory argument which takes the instance and while printing we can just access the attribute value and print it. Due to which everytime we call the method we dont need to pass argument.

````python
#First example
class Fruits:
   def greet(self,name):
      print(f"I am {name}")

apple = Fruits()
apple.greet("apple")
````

````python
#Second example
class Fruits:
   def __init__(self,name):
     self.name = name
   def greet(self):
      print(f"I am {self.name}")

apple = Fruits("apple")
apple.greet()
````
