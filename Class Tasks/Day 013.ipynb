#**Types of Attributes**
1. Instance attribute

    These are the attributes that belong to instances of a class. They are defined within the constructor method `__init__` and can be accessed using the `self` keyword like `self.name` and `self.age` in the above Dog class. They are initialized when a new instance of the class is created.


2. Class attributes

    Class attributes are attributes that belong to the class itself. They are defined outside the constructor method `__init__` and can be accessed using the class name. Class attributes are shared by all instances of the class.


````python
#Instance Attribute examples
class Student:
  ''' Student Information''' #docstring
  def __init__(self,name):
    self.name = name #Here in instance attribute all the value goes to self and be stored
                     #Here the attribute is defined inside the init function while initializing

s1 = Student("Maradona")
print(s1.name)
````

So there might be question if it can work without `@staticmethod` decorator why we should use it so here is the answer:


*   Without `@staticmethod`:

    If you define a plain method inside a class (without `self` or `cls`), Python still treats it as an instance method by default.

    That means if you call it through an instance (`obj.method()`), Python automatically passes the instance (`self`) as the first argument — which will break if your method doesn’t expect it.

    If you call it through the class (`ClassName.method()`), Python doesn’t inject self, so it works only because you’re bypassing the instance mechanism.

* With @staticmethod:

    You’re explicitly telling Python: “This method does not depend on the instance (`self`) or the class (`cls`). Treat it like a plain function that just happens to live inside the class namespace.”

    Whether you call it via `ClassName.method()` or `obj.method()`, Python will not pass any extra arguments. It behaves exactly like a normal function.

````python
#Example
class Demo:
    def normal_method(x):   # no self
        print(x)

    @staticmethod
    def static_method(x):
        print(x)

# Calling through the class
Demo.normal_method(10)   # ✅ Works
Demo.static_method(10)   # ✅ Works

# Calling through an instance
obj = Demo()
obj.static_method(20)    # ✅ Works (no self passed)
try :
  obj.normal_method(20)    # ❌ TypeError: got 2 arguments, expected 1
except TypeError:
  print("TypeError: normal_method() missing 1 required positional argument: 'x'")
````

⚖️ Why Use @staticmethod?
* Clarity: It signals to readers that the method doesn’t touch instance or class data.

* Consistency: It avoids accidental self injection when called via an object.

* Organization: It keeps utility/helper functions grouped inside the class, even though they don’t depend on it.

So while both `ClassName.method()` calls look the same, the difference shows up when you call through an instance. `@staticmethod` makes the intent explicit and prevents Python from trying to sneak in `self`

Lets summarize the methods then:

1. If you want to pass the `instance` as argument or use instance attributes/variables then use `instance method` which is `normal method` inside class or `__init__`.
   * It always has one positional attribute where `self ` (meaning instant itself) is written as the standard name for that attribute which takes `instance`.

2. If you want to pass the `class` as argument or use class attributes/variables then use `classmethod` which is created by the programmer using `@classmethod` decorator above the method they want to create as classmethod

   * It always has one positional attribute where `cls ` (meaning class) is written as the standard name for that attribute which takes `class`.


3. If you don't want to pass the `class` or `instance` as argument or then use `staticmethod` which is created by the programmer using `@staticmethod` decorator above the method they want to create as staticmethod

   * It doesn't compulsorily require positional attribute.

````python
class Dog:
  count = 0 #class attribute

  def __init__(self,name,age = 3):
    self.name = name
    self.age = age
    Dog.count += 1

  @classmethod
  def get_count(cls):
    return cls.count

  def bark(self):
    print(f"{self.name} is barking")

  @staticmethod
  def get_vet_name():
    return "Dr. Ram K.C"


d = Dog("Tommy",21)
print(d.name)
print(d.age)
d.bark()
print(Dog.get_count())
print(d.get_count())
print(d.get_vet_name())
print(Dog.get_vet_name())

d2 = Dog("Jimmy")
print(d2.name)
print(d2.age)
d2.bark()
print(Dog.get_count())
print(d2.get_count())
print(d2.get_vet_name())
print(Dog.get_vet_name())
````

Student Grade Management System
Create a Python class named Student.
Requirements:
1. Create a class attribute named school.
2. Use the __init__() method to take:
    * Student name
    * Student marks
3. Create an instance method named check_grade() that displays:

    A Grade: Marks ≥ 80

    B Grade: Marks ≥ 60 and < 80

    C Grade: Marks ≥ 40 and < 60

    Fail: Marks < 40
    
4. Create a class method to display the school name.
5. Create a static method to display a welcome message.
6. Create an object and display the student's grade

````python
class Student:
  school = "MeroSchool"
  def __init__(self):
    self.name = input("Enter student name: ")
    self.marks = float(input("Enter student marks: "))

  def check_grade(self):
    if self.marks >= 80:
      print("\nResult: A Grade")
    elif self.marks >= 60 and self.marks < 80:
      print("\nResult: B Grade")
    elif self.marks >= 40 and self.marks < 60:
      print("\nResult: C Grade")
    else:
      print("\nResult: Fail")

  @classmethod
  def show_school(cls):
    print(f"School Name: {cls.school}\n")

  @staticmethod
  def welcome():
    print("Welcome to Student Grade Management System\n")

Student.show_school()
Student.welcome()
S1 = Student()
S1.check_grade()
````

    * Account holder name
    * Initial balance
3. Create the following instance methods:
   * deposit() → Ask the user for an amount and add it to the balance.
   * withdraw() → Ask the user for an amount and deduct it from the balance if sufficient balance is available.
   * show_balance() → Display the current balance.
4. Create a class method to display the bank name.
5. Create a static method to display banking rules or a welcome message.
6. Create an object and test all methods.

````python
class BankAccount:
  bank_name = "MeroBank"
  def __init__(self):
    self.name = input("Enter account holder name: ")
    self.balance = float(input("Enter initial balance: "))
  def deposit(self):
    amount = float(input("\nEnter amount to deposit: "))
    self.balance += amount
    print(f"\nNew balance: {self.balance}")
  def withdraw(self):
    amount = float(input("\nEnter amount to withdraw: "))
    if amount > self.balance:
      print("\nInsufficient balance")
    else:
      self.balance -= amount
      print(f"\nNew balance: {self.balance}")
  def show_balance(self):
    print(f"\nCurrent balance: {self.balance}")

  @classmethod
  def show_bank_name(cls):
    print(f"Bank Name: {cls.bank_name}\n")

  @staticmethod
  def welcome():
    print("Welcome to Bank Account Management System\n")



BankAccount.show_bank_name()
BankAccount.welcome()
B1 = BankAccount()
B1.deposit()
B1.withdraw()
B1.show_balance()
````
