```python
#dictionary --> variable = {"key1" : value1 , "key2" : "value2"}
#example
dat = {
    "id" : 101,
    "name" : "Damon salvatore",
     "cont" : {
         "email" : "damon@yahoo.com"
     }
    }
print(type (dat))
```

```python
#creation of dictornary using dict() constructor
#syntax --> dict({key:value,key2:value2})
dat1 = dict({"game":"fallguys","price":"2022"})
dat2 = dict({'fifa':2026})
print(type (dat1))
print(type (dat2))
```

```python
#nested dictionary
#using constructor
dat = dict({'student1': dict({'name': 'Dhiraj', 'age': 22})})
print(dat)

#using normal way
dat1 = {'student2': {
    'name': 'idk', 'age': 22}
    }
print(dat1)
```

```python
# create a dictionary from sequence having each item as a pair
#using constructor like dict({(key1,value1),(key2,value2)...})
person = dict(
    {("name", "rita"),
               ("country", "nepal")}
              )
print(person)
```

```python
# create dictionary with mixed keys keys
sample_dict = {"name": "gita", 10: "Mobile"}
print(sample_dict)
```

```python
#create dictionary with value as a list
person = {"name": "sita", "telephones": [9898000001, 9898000002,9898000000]}
print(person)
```

```python
#get specific value
# dictionary_name.get(key_name)
sample = {"name": "gita", "device": "Mobile"}
print(sample.get("device"))
#or
# dictionary_name[key_name]
print(sample["name"])
```

```python
dat = {'student1': {'name': 'Dhiraj', 'age': 22},
       'student2': {'name': 'Ram', 'age': 21}}


print(dat.keys(),"\n") #gives all the keys
print(dat.values(),"\n") #gives all the values
print(dat.items(),"\n") #gives everything ie key and value
print(dat.get( 'student1' ),'\n') #gives value of specific key
dat['student1']['name']='hello world' #changes the value of the nested key
                                      #mentioned could be used to create the key
                                      #value pair if key of same name is not found

print(dat,'\n')                       #prints the dictionary
print(dat['student2']['name'],'\n')   #gives the value of specific nested key

```

```python
#more functions for dictionary

dat = {
    "name" : "tom cruise",
    "age" : 21,
    "faculty" : "movie",
    "city" : "LA",
    12 : 45
}
print(dat)
dat.update({"age":50})  #only one argument and changes the value
print(dat)
dat.update(dict({"phone":9898000000})) #new key value pairs can be added this way
print(dat)

del(dat["city"]) #deletes key
print(dat)
try:
  dat.key(["city"]) # gives attributeerror if key is not found
except AttributeError:
  print("error key not found")
```

```python
#sorting
student = {
    'b' : 1,
    'd' : 4,
    ' c' : 2,
    'a' : 3
}
print(sorted(student)) #look it sorts key so the key are in alphabets so it
                      # sorted key in order of alphabet
                      # and it takes empty value in first of order
student1 = {
    2 : 'a',
    1 : 'b',
    6 : 'c',
    4 : 'd'

}
print(sorted(student1)) #look here the key is sorted in ascending order
```
