```python
#no argument no return
def sum():
  c = int(input("Enter a number: \n"))
  d = int(input("Enter a number: \n"))
  s = c + d
  print("The sum is: ",s)
sum()
```

```python
#yes argument no return
def sum(a,b):
  s = a + b
  print("The sum is: ",s)
c = int(input("Enter a number: \n"))
d = int(input("Enter a number: \n"))
sum(c,d)
```

```python
#no argument yes return
def sum():
  c = int(input("Enter a number: \n"))
  d = int(input("Enter a number: \n"))
  s = c + d
  return s
e = sum()
print("The sum is: ",e)
```

```python
#yes argument yes return
def sum(a,b):
  s = a + b
  return s

c = int(input("Enter a number: \n"))
d = int(input("Enter a number: \n"))
print("The sum is: ",sum(c,d))
```

```python
#greeting function
def greet(a):
  print("Good Morning",a)
a = input("Enter your name: ")
greet(a)
```

```python
#calculator code using function
def calc(op, n1, n2):
  if op == '+':
    return n1 + n2
  elif op == '-':
    return n1 - n2
  elif op == '*':
    return n1 * n2
  elif op == '/':
    if n2 == 0:
      print("Undefined")
    else:
      return n1 / n2
  else:
    return
n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))
op = input("Enter an operator: ")
if op != '+' and op != '-' and op != '*' and op != '/': #we can also use this
                        #not in ['+', '-', '*', '/']:
                         #(after if statement in place of current code)
  print("Invalid operator")
else:
  print("The answer is: ", calc(op, n1, n2))
```

```python
#function to return area of rectangle
def area(length,breadth):
  a = length * breadth
  return a
length = int(input("Enter the length: "))
breadth = int(input("Enter the breadth: "))
print("The area of rectangle is: ",area(length,breadth))
```
