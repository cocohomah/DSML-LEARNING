# **Exception Handling**
When we get errors like TypeError, ValueError (the error which is in red color when running the program), it can interrupt the program to further continue because of which it can degrade the user experience.

So to prevent that but only display the error msg and continue or to do something else when error is raised, we use exception handling.

**Syntax for error handling:**
```
try:
 Statements
except Error_Name:
 statements
finally:
 statements
```
**Notes:**

 * In except, error_name is not compulsory.
 * In finally, we keep the command which will be executed after the  exceution of exception handling.
 * `finally`'s statement will always be exceuted either its after `try` or `except`


````python
#Examples
#ZeroDivisionError
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
try:
  result = num1/num2
  print("The division is: ",result)
except ZeroDivisionError:
  print("It can't be divided since num2 is 0")
````

````python
#TypeError
num1 = int(input("Enter the first number: "))
num2 = input("Enter the second number: ")
try:
  result = num1+num2
  print("The addition is: ",result)
except TypeError:
  print("It can't be added because num1 is integer and num2 is string so its a TypeError")
````

````python
#ValueError
num1 = int(input("Enter the first number: "))
try:
  num2 = int(input("Enter the second number: "))
  result = num1+num2
  print("The addition is: ",result)
except ValueError:
  print("It is giving ValueError because A is given as num2 than some integer value")
````

````python
#IndexError
lst = [0,1,2,3]
n = int(input("Enter the index: "))
try:
  print(lst[n])
except IndexError:
  print(f"IndexError is given because index {n} doesnt exist in list")
````

````python
#KeyError
dic = {
    "name": "AquaMan"
}
try:
  print(dic["age"])
except KeyError:
  print("KeyError is given because 'age' key doesnt exist in dictionary")
````

````python
#FileNotFound Error
try:
  f = open("hello.txt","r")
except FileNotFoundError:
  print("FileNotFoundError is given because hello.txt file doesnt exist")
````

````python
#finally used after Try
try:
  print(2+2)
except:
  print("Something went wrong")
finally:
  print("The finally block is executed")
````

# **USER RAISED ERROR**

This is done when there is no built-in error / we want a custom error to be displayed when running a function or a program.


**Syntax for raising a error :**
```
raise ERROR_NAME("The message to print when error is showed")
```

````python
#USER_RAISED_ERROR Example
#creating error using built-in error name
def check(num):
  if num<0:
    raise ValueError("The number cant be negative")
  print("Number: ",num)


check(5)
print("\nNow using negative number(-5)\n")
try:
  check(-5)
except ValueError as e: #here message is recceived and stored in e
  print(e) #here message is displayed
````

````python
#creating error using custom name

class Error(Exception):   #here we are creating a error class with exception parent (which should be used for creating all error classes)
                          #both built-in and user-defined uses this parent class to create error class
  pass

def check(num):
  if num<0:
    raise Error("The number cant be negative")
  print("Number: ",num)


check(5)
print("\nNow using negative number(-5)\n")
try:
  check(-5)
except Error as e: #here message is recceived and stored in e
  print(e) #here message is display
````

````python
#raise error when student marks is not >0 and <100
def check(num):
  if num <0 or num >100:
    raise ValueError("The marks should be between 0 and 100")
  print("Marks: ",num)

marks = float(input("Enter the marks of the student: \n"))
try:
  check(marks)
except ValueError as e:
  print(e)
````

````python
#if age is negative
def check(age):
  if age<0:
    raise ValueError("Age cant be negative")
  print("Age: ",age)

age = int(input("Enter the age: \n"))
try:
  check(age)
except ValueError as e:
  print(e)
````

````python
#if password is less than 8 character long
class InvalidPasswordError(Exception):
  pass

def check(password):
  if len(password)<8:
    raise InvalidPasswordError("Password should be atleast 8 characters long")
  print("Password: ",password)

password = input("Enter the password: \n")
try:
  check(password)
except InvalidPasswordError as e:
  print(e)
````

````python
class InsufficientAMT(Exception):
 """The amount is insufficient."""
 pass
class NegativeNumberError(Exception):
 """The number is negative."""
 pass

amount_balance = 1000


def check_withdraw(n,amount_balance = amount_balance):
   if n > amount_balance:
    raise InsufficientAMT
   elif n<0:
    raise NegativeNumberError


def check_deposit(n):
   if n<0:
    raise NegativeNumberError


def deposit():
  global amount_balance   # here this is amount_balance is a global variable
                          #which means hey function when u change smth to this
                          #variable it should be changed in main place of code(global scope) too
                          #so now python treats it as global variable than local one
  try:
   n = float(input("\nEnter the amount to be deposited: "))
   check_deposit(n)
   amount_balance += n
   print("\nThe amount is added")
   print("\nThe updated amount is: ", amount_balance)

  except NegativeNumberError:
   print("\nThe amount cant be negative\nPlease Try Again!\n")
   # Call deposit again to re-prompt for valid input
   deposit()
  except ValueError:
   print("\nInvalid input\nPlease Try Again!\n")
   # Call deposit again to re-prompt for valid input
   deposit()
  except TypeError:
   print("\nInvalid input\nPlease Try Again!\n")
   # Call deposit again to re-prompt for valid input
   deposit()


def withdraw( ):
    global amount_balance
    try:
        n = float(input("\nEnter the amount to be withdrawn: "))
        check_withdraw(n)
        amount_balance -= n
        print("\nThe amount is withdrawn")
        print("\nThe updated amount is: ", amount_balance)

    except InsufficientAMT:
        print("\nThe amount is more than account balance\nPlease Try Again!\n")
        # Call withdraw again to re-prompt for valid input
        withdraw()

    except NegativeNumberError:
        print("\nThe amount cant be negative\nPlease Try Again!\n")
        # Call withdraw again to re-prompt for valid input
        withdraw()

    except ValueError:
        print("\nInvalid input\nPlease Try Again!\n")
        # Call withdraw again to re-prompt for valid input
        withdraw()

    except TypeError:
        print("\nInvalid input\nPlease Try Again!\n")
        # Call withdraw again to re-prompt for valid input
        withdraw()



def balance():
  global amount_balance
  try:
   print("\nYour account balance is: ")
  finally:
     print(amount_balance,"\n")




print("Welcome to the ATM\n\n")
while True:
  num = int(input("\nENTER\n1-> Deposit\n2-> Withdraw\n3-> Balance\n4-> Exit\n\n"))
  if num == 1:
    deposit()
  elif num == 2:
    withdraw()
  elif num == 3:
    balance()
  elif num == 4:
    print("Thank you for using the ATM")
    break
  else:
    print("Invalid input")

  while True:
    choice = input("\nDo you want to continue? (y/n): ")
    if choice.lower().strip() != "y" and choice.lower().strip() != "n":
      print("Invalid input\nPlease Try Again!\n")
    else:
      break
  if choice == "n":
    print("Thank you for using the ATM")
    break
````
